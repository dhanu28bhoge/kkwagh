```python

import os
import re

pages = {}

# Read HTML files
for file in os.listdir("website"):

    if file.endswith(".html"):

        f = open("website/" + file, encoding="utf-8")
        text = f.read().lower()
        f.close()

        # Remove HTML tags
        text = re.sub("<.*?>", "", text)

        pages[file] = text

print("Search Engine Ready!")

# Take input from user
query = input("Enter keyword: ").lower()

found = False

# Search keyword
for file in pages:

    if query in pages[file]:
        print("Found in:", file)
        found = True

if found == False:
    print("No result found")
```
