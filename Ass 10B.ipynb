```python

import requests
import csv
from bs4 import BeautifulSoup
from urllib.parse import urljoin

url = "https://books.toscrape.com/catalogue/page-1.html"
data = []

for i in range(3):

    response = requests.get(url, timeout=20)
    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")

    books = soup.find_all("article", class_="product_pod")

    for b in books:

        name = b.h3.a["title"]
        price = b.find("p", class_="price_color").text
        stock = b.find("p", class_="availability").text.strip()
        link = urljoin(url, b.h3.a["href"])

        data.append([name, price, stock, link])

    next_page = soup.find("li", class_="next")

    if next_page:
        url = urljoin(url, next_page.a["href"])
    else:
        break

with open("products.csv", "w", newline="", encoding="utf-8") as f:

    w = csv.writer(f)
    w.writerow(["Name", "Price", "Stock", "Link"])
    w.writerows(data)

print("Crawling Completed!")
print("Total Products:", len(data))
print("Saved in products.csv")
```
