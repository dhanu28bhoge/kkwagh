```python

import re
from nltk.stem import PorterStemmer

# Input document
document = """
Information Retrieval systems retrieve relevant documents from large collections.
Retrieving, retrieved, and retrieval are related terms.
"""

# Step 1: Tokenization
tokens = re.findall(r'\b[a-z]+\b', document.lower())

# Step 2: Remove stopwords
stop_words = {
    "from", "and", "are", "the", "is", "a", "an", "of"
}

words = [w for w in tokens if w not in stop_words]

print("After removing stopwords:")
print(words)

# Step 3: Stemming
stemmer = PorterStemmer()

conflated = [stemmer.stem(w) for w in words]

print("\nConflated Terms:")
print(conflated)

# Step 4: Document Representation
representation = {}

for w in conflated:
    representation[w] = representation.get(w, 0) + 1

print("\nDocument Representation:")
print(representation)
```
